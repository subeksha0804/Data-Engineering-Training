```python
import dlt
from pyspark.sql.functions import *

@dlt.table(
    name="bronze_patient_visits",
    comment="Raw patient visit data"
)
def bronze_patient_visits():
    data = [
        (26,1002,201,"2024-03-14","Completed",2),
        (27,1005,210,"2024-03-14","Completed",1),
        (11,1011,205,"2024-03-06","Completed",3),
        (20,1020,206,"2024-03-10","Completed",2),
        (28,1018,203,"2024-03-15","Pending",2)
    ]

    columns = [
        "visit_id","patient_id","doctor_id",
        "visit_date","visit_status","tests_count"
    ]

    return spark.createDataFrame(data, columns)
```

```python
@dlt.table(
    name="silver_patient_visits",
    comment="Cleaned patient visits"
)
def silver_patient_visits():
    return dlt.read("bronze_patient_visits") \
        .withColumn("visit_date", to_date("visit_date", "yyyy-MM-dd"))
```

```python
@dlt.table
def silver_patient_visits_enriched():
    return dlt.read("silver_patient_visits") \
        .withColumn("test_cost", col("tests_count") * 500) \
        .withColumn("consultation_fee", lit(1000)) \
        .withColumn("total_bill", col("test_cost") + col("consultation_fee"))
```

```python
@dlt.table
def silver_cleaned():
    return dlt.read("silver_patient_visits_enriched") \
        .filter(col("visit_id").isNotNull()) \
        .filter(col("visit_status").isin("Completed","Pending"))
```

```python
@dlt.table(
    name="gold_city_revenue"
)
def gold_city_revenue():
    return dlt.read("silver_cleaned") \
        .withColumn("city", lit("Unknown")) \
        .groupBy("city") \
        .agg(sum("total_bill").alias("total_revenue"))
```

```python
@dlt.table(
    name="gold_specialization_revenue"
)
def gold_specialization_revenue():
    return dlt.read("silver_cleaned") \
        .withColumn("specialization", lit("General")) \
        .groupBy("specialization") \
        .agg(sum("total_bill").alias("total_revenue"))
```
